```python
from arcgis.gis import GIS
from arcgis.geocoding import geocode
from arcgis.geoenrichment import BufferStudyArea, enrich
from arcgis.geometry import Geometry

gis = GIS(username='gpec447sp26_33')
```

```python
from pathlib import Path
import geopandas as gpd

# Find project folder from current notebook location
current_dir = Path.cwd()

# Data folder
data_dir = current_dir / "data"
```

```python
lrt_stops_path = data_dir / "lrt_stops.shp"

lrt_stops = gpd.read_file(lrt_stops_path)

lrt_stops.head()
```

```python
shp_files = list(data_dir.glob("lrt_stops.shp"))
shp_files
```

```python
lrt_stops_path = shp_files[0]   # change index if needed
lrt_stops = gpd.read_file(lrt_stops_path)

lrt_stops.head()
```

```python
from arcgis.geometry import Geometry
from arcgis.geoenrichment import BufferStudyArea

study_areas = []

# Make sure lrt_stops is in WGS84 for ArcGIS API
lrt_stops_wgs84 = lrt_stops.to_crs(epsg=4326)

for idx, row in lrt_stops_wgs84.iterrows():
    pt = Geometry({
        "x": row.geometry.x,
        "y": row.geometry.y,
        "spatialReference": {"wkid": 4326}
    })

    area = BufferStudyArea(
        area=pt,
        radii=[5, 10, 15],
        units="Minutes",
        travel_mode="Walking",
        overlap=True
    )

    study_areas.append(area)
```

```python
from arcgis.geoenrichment import enrich

walk_results = enrich(
    study_areas=study_areas,
    data_collections=["KeyGlobalFacts"]
)

walk_results.head()
```

Here it shows areas that are farther from the LRT stop in different travel times (5 min, 10 min, 15 min).

```python
m = gis.map("Utsunomiya, Japan")

walk_results.spatial.plot(
    map_widget=m,
    renderer_type="c",          # class breaks / graduated colors
    col="buffer_radii",          # column for 5, 10, 15 minute areas
    class_count=3,
    alpha=0.45,
    legend=True
)

m
```
